# 00 — Overview: ablation & comparison study across all 5 targets

This directory answers one question per target, with evidence: **why was this specific
architecture chosen over every reasonable alternative?** It consolidates this project's full
experiment history (classical ML baselines, every alternative neural architecture actually
tried -- attention, ChemBERTa embeddings, graph neural networks, evidential regression -- and
every intermediate step of the winning architecture's own development) into sourced comparison
tables and publication-format figures.

*Every number in this notebook is read from an existing result file already produced by this
project (see the `source` column) -- this notebook does not train any model. The `Classical`,
`MFMN_Attention`, `ChemBERTa`, `GraphMPNN`, `Evidential`, and intermediate `MFMN_DynGate`/
`MFMN_InteractAux` rows are historical experiment-log results; the row marked **FINAL** was
independently re-verified this session (a fresh rerun matched the historical number closely --
see `mfmn_benchmark_package/notebooks/` for that from-scratch, fully executed reproduction).*


## Headline result, all 5 targets

| Target | Chosen model | Best alternative | R² gap | Why the alternative lost |
|---|---|---|---|---|
| pKa_Acidic | GraphMPNN (GNN) | MFMN_Attention | 0.969 vs 0.942 | No site-local representation |
| pKa_Basic | GraphMPNN (GNN) | Classical RF+SVM+XGB | 0.948 vs 0.882 | No site-local representation |
| CL | MFMN_DynGate | Classical RF+SVM+XGB | 0.503 vs 0.285 (official-test) | CV didn't predict test; GNN underfits |
| VDss | MFMN_InteractAux | MFMN_DynGate (CV) | 0.641 vs 0.454 (CV, different protocol) | Gating didn't help this target; simplicity wins |
| Fu | MFMN_DynGate + 1.25x weighting | Classical RF+SVM+XGB | 0.712 vs 0.660 | No localized site for GNN; classical competitive but behind |

## Files in this directory

- `data/all_models_comparison.csv` — every number, every model, every target, with exact source file
- `data/build_comparison_tables.py` — regenerates the CSV from the underlying result files
- `data/make_figures.py` — regenerates all 5 comparison figures
- `figures/*_comparison.png` — one two-panel (R² + headline error metric) figure per target
- `notebooks/01-05_*.ipynb` — one notebook per target with the full table + figure + narrative

## Reading the figures

Bars are colored by model family (same color = same family across every target's chart).
**Hatched bars are cross-validation estimates; solid bars are one-shot official test-set
results** — these two protocols are not directly comparable to each other (see the CL notebook
for a case where this actually matters: CV rankings did not predict the official-test winner).
The chosen model's bar has a bold black outline. The dashed vertical line is the published
paper's benchmark value for that metric.
